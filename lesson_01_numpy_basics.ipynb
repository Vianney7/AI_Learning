```python
# Lesson 1: NumPy = how AI does math
import numpy as np

# Think of this like an Excel row, but for AI
data = np.array([10, 10, 10, 10, 10])
print("Our data:", data)

# AI models love matrix math. This is a "dot product" - super common
weights = np.array([0, 0, 1, 0, 0])
prediction = np.dot(data, weights)

print("Prediction from our tiny 'model':", prediction)
print("Data shape:", data.shape)  # AI people obsess over shapes
print("Sum of data:", np.sum(data))
print("Mean of data:", np.mean(data))
print("Prediction with 0.2 weights:", np.sum(data) * 0.2)
```

```python
# Lesson 2: Pandas — Real Data

import pandas as pd

# Pandas can read data straight from the internet
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)

print("First 5 rows:")
print(df.head())

print("\nShape of dataset:", df.shape)  # (rows, columns)
print("\nColumn names:", df.columns.tolist())
print("\nAverage tip:", df['tip'].mean())
```
