```python
import random

movies = [
    {
        'title': 'The Matrix',
        'genre': 'Syfy',
        'rating': 9.0    
    },
    {
        'title': 'The Avengers',
        'genre': 'Action',
        'rating': 8.5
    },
    {
        'title': 'The Notebook',
        'genre': 'Romance',
        'rating': 7.0
    },
    {
        'title': 'Avatar',
        'genre': 'Adventure',
        'rating': 9.5
    },
    {
        'title': 'Terminator 2: Judgment Day',
        'genre': 'Sci-Fi',
        'rating': 8.8
    }
]

must_watch = []
for movie in movies:
    if movie['rating'] >= 8.0:
        must_watch.append(movie)

suggestion = random.choice(must_watch)
print('Tonight you should watch: ' + suggestion['title'])    
```
