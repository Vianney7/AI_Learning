```python
import matplotlib.pyplot as plt
import numpy as np

# ========================
# World Development Data 2007
# ========================

# GDP per capita (in USD)
gdp_cap = [
    500, 800, 1200, 1550, 3000,
    4500, 5700, 7500, 9000, 11000
]

# Life expectancy (in years)
life_exp = [
    42, 48, 52, 71, 60,
    63, 80, 68, 70, 72
]

# Population (in millions)
pop = [
    5, 8, 12, 1200, 25,
    35, 1400, 50, 61, 75
]

# Colors for each country
col = [
    'red', 'blue', 'green', 'yellow', 'red',
    'blue', 'green', 'yellow', 'black', 'red'
]

# ========================
# Create the Scatter Plot
# ========================

plt.figure(figsize=(10, 6))

plt.scatter(
    x=gdp_cap,
    y=life_exp,
    s=np.array(pop) * 2,      # Bubble size proportional to population
    c=col,
    alpha=0.8,
    edgecolors='black',
    linewidth=0.5
)

# Logarithmic scale on x-axis (common for GDP)
plt.xscale('log')

# Add labels and title
plt.xlabel('GDP per Capita [in USD]', fontsize=12)
plt.ylabel('Life Expectancy [in years]', fontsize=12)
plt.title('World Development in 2007', fontsize=14)

# Add grid for better readability
plt.grid(True, which="both", ls="--", alpha=0.5)

plt.show()
```

```python
# Country names (matching the order of the lists)
countries = ['India', 'Nigeria', 'Pakistan', 'China', 'Indonesia',
             'Brazil', 'United States', 'Mexico', 'Russia', 'Japan']

# Enhanced scatter plot with labels
plt.figure(figsize=(12, 7))

scatter = plt.scatter(
    x=gdp_cap,
    y=life_exp,
    s=np.array(pop) * 2,
    c=col,
    alpha=0.8,
    edgecolors='black'
)

plt.xscale('log')
plt.xlabel('GDP per Capita [in USD]', fontsize=12)
plt.ylabel('Life Expectancy [in years]', fontsize=12)
plt.title('World Development in 2007', fontsize=14)
plt.grid(True, ls="--", alpha=0.5)

# Add country labels
for i, country in enumerate(countries):
    plt.annotate(country,
                 (gdp_cap[i], life_exp[i]),
                 xytext=(5, 5),
                 textcoords='offset points',
                 fontsize=9)

plt.show()
```

```python
import pandas as pd

# Create sample dataset
cars_dict = {
    'country': [
        'United States',
        'Australia',
        'Japan',
        'India',
        'Russia',
        'Morocco',
        'Egypt'
    ],
    'drives_right': [
        True, False, False,
        False, True, True, True
    ],
    'cars_per_cap': [
        809, 731, 588,
        18, 200, 70, 45
    ]
}

# Create DataFrame
cars = pd.DataFrame(cars_dict)

# Add row labels
cars.index = [
    'US', 'AUS', 'JPN',
    'IN', 'RU', 'MOR', 'EG'
]

# Print country column as Series
print(cars['country'])

print()

# Print country column as DataFrame
print(cars[['country']])

print()

# Print multiple columns
print(cars[['country', 'drives_right']])
```
