```python
import matplotlib.pyplot as plt
import numpy as np

# GDP per capita
gdp_cap = [
    500, 800, 1200, 2000, 3000,
    4500, 6000, 7500, 9000, 11000
]

# Life expectancy
life_exp = [
    42, 48, 52, 57, 60,
    63, 66, 68, 70, 72
]

# Population (millions)
pop = [
    5, 8, 12, 18, 25,
    35, 42, 50, 61, 75
]

# Convert list to NumPy array
np_pop = np.array(pop)

# Double bubble sizes
np_pop = np_pop * 2

# Scatter plot with bubble sizes
plt.scatter(gdp_cap, life_exp, s=np_pop)

# Log scale
plt.xscale('log')

# Labels and title
plt.xlabel('GDP per Capita [in USD]')
plt.ylabel('Life Expectancy [in years]')
plt.title('World Development in 2007')

# Custom x-axis ticks
plt.xticks(
    [1000, 10000, 100000],
    ['1k', '10k', '100k']
)

# Show plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# GDP per capita
gdp_cap = [
    500, 800, 1200, 2000, 3000,
    4500, 6000, 7500, 9000, 11000
]

# Life expectancy
life_exp = [
    42, 48, 52, 57, 60,
    63, 66, 68, 70, 72
]

# Population (millions)
pop = [
    5, 8, 12, 18, 25,
    35, 42, 50, 61, 75
]

# Colors by continent
col = [
    'red', 'blue', 'green', 'yellow', 'red',
    'blue', 'green', 'yellow', 'black', 'red'
]

# Scatter plot
plt.scatter(
    x=gdp_cap,
    y=life_exp,
    s=np.array(pop) * 2,
    c=col,
    alpha=0.8
)

# Logarithmic x-axis
plt.xscale('log')

# Labels and title
plt.xlabel('GDP per Capita [in USD]')
plt.ylabel('Life Expectancy [in years]')
plt.title('World Development in 2007')

# Custom ticks
plt.xticks(
    [1000, 10000, 100000],
    ['1k', '10k', '100k']
)

# Display plot
plt.show()
```
